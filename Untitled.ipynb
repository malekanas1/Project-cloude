```python
def factorial(n):
    if n < 0:
        return "المضروب غير معرف للأعداد السالبة"
    elif n == 0:
        return 1
    else:
        fact = 1
        for i in range(1, n + 1):
            fact = fact * i
        return fact

# مثال على الاستخدام:
number = 5
result = factorial(number)
print(f"مضروب العدد {number} هو: {result}")

# يمكنك تغيير قيمة 'number' لاختبار أعداد أخرى
number_two = 7
result_two = factorial(number_two)
print(f"مضروب العدد {number_two} هو: {result_two}")

```
